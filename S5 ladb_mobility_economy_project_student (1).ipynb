### 2.2 Renombrar columnas

```python
# Estandarizar los nombres de las columnas de traffic
traffic = traffic.rename(columns={'Country':'country', 'UpdateTimeUTC': 'update_time_utc', 'UpdateTimeUTCWeekAgo': 'update_time_utc_week_ago', 'City':'city'})

# verificar cambios
traffic.columns
```

```python
# Estandarizar los nombres de las columnas de eco
eco = eco.rename(columns={'Country':'country','City GDP/capita': 'city_gdp_capita',
    'Unemployment %': 'unemployment_pct',
    'Population (M)': 'population_m', 
    'City':'city', 
    'PM2.5 (μg/m³)':'pm25',
    'Year': 'year'})
# verificar cambios
eco.columns
```


### 2.3 Corregir formatos numéricos y de fecha

```python
# Convertir las columnas de traffic a tipo fecha con pd.to_datetime()
traffic['update_time_utc'] = pd.to_datetime(traffic['update_time_utc'], errors='coerce', utc= True)
traffic['update_time_utc_week_ago'] =  pd.to_datetime(traffic['update_time_utc_week_ago'], errors='coerce', utc= True)

# verificar el cambio
traffic.info()
```

```python
# Limpia separadores y convierte columnas numéricas en eco
eco['city_gdp_capita'] = eco['city_gdp_capita'].astype(str).str.replace('.', '').str.replace(',', '.').astype(float)
eco['unemployment_pct'] = eco['unemployment_pct'].astype(str).str.replace('%', '').str.replace(',', '.').astype(float)
eco['population_m'] = eco['population_m'].astype(str).str.replace(',', '.').astype(float)

# Calcula la población total en unidades absolutas (Multiplica * 1000000)
eco['population'] = eco['population_m'] * 1000000

# verificar el cambio
eco.info()
eco.head(3)
```


---

## 🧩Paso 3: Extraer año y filtrar

Extraer el año permite filtrar la información y trabajar solo con el período más reciente y relevante.

### 3.1 Extraer columna año y filtrar 2024

```python
# Extraer el año de las fechas en update_time_utc
traffic['year'] = traffic['update_time_utc'].dt.year

# Verificar cambio
traffic.head(3)
```

```python
# Filtra los registros del año 2024
traffic_2024 = traffic[traffic['year'] == 2024].copy()
eco_2024 = eco[eco['year'] == 2024].copy()

# Revisar dataframes nuevos
display(traffic_2024.head())
display(eco_2024.head())
```


---

## 🧩Paso 4: Analizar y resumir datos de movilidad

Como el dataset de tráfico contiene **múltiples registros por ciudad**. En esta parte, se calculan los promedios anuales por ciudad para simplificar el análisis y obtener una visión más clara de las tendencias generales.

### 4.1 Calcular promedios de tráfico por ciudad

```python
# Calcular los  promedios de trafico por ciudad, país y año
traffic_city_year_2024 =  traffic_2024.groupby(['city','country','year'])[['JamsDelay','TrafficIndexLive', 'JamsLengthInKms', 'JamsCount', 'MinsDelay', 'TravelTimeLivePer10KmsMins', 'TravelTimeHistoricPer10KmsMins']].agg('mean').reset_index()

# Mostrar resultado
traffic_city_year_2024.head(10)
```

```python
traffic_city_year_2024.sort_values(['JamsDelay'], ascending=False)
```

La ciudad con el mayor tiempo promedio de tráfico es Ciudad de México


---

## 🧩Paso 5: Unir movilidad y economía

Combinar datasets para analizar cómo se relacionan los indicadores económicos con los de movilidad.

### 5.1 Unir tráfico (tabla principal) con indicadores económicos

```python
# Seleccionar columnas clave de tráfico y economía
left_cols = ['city','country','year','JamsDelay','TrafficIndexLive', 'JamsLengthInKms', 'JamsCount', 'MinsDelay', 'TravelTimeLivePer10KmsMins', 'TravelTimeHistoricPer10KmsMins']


right_cols = ['city','year','city_gdp_capita','unemployment_pct','pm25','population']

# Usar .copy() para crear los dos nuevos datasets reducidos
traffic_2024_small = traffic_city_year_2024[left_cols].copy()
eco_2024_small = eco_2024[right_cols].copy()

# Unir datasets
merged = pd.merge(traffic_2024_small, eco_2024_small, on=['city', 'year'], how='inner')

# Mostrar las primeras 5 filas
merged.head(5)
```


---

## 🧩Paso 6: Visualización y análisis de relaciones
Los gráficos a continuación permitirán entender cómo se relacionan las variables económicas con las de movilidad urbana para **visualizar patrones o tendencias**.

### 6.1 Visualizar relaciones entre economía y tráfico

```python
# Crear boxplot para observar el comportamiento de los minutos de congestion JamsDelay
sns.boxplot(y = merged['JamsDelay'], showmeans=True)

# obtener promedio para mostrarlo en título
mean_value = merged['JamsDelay'].mean()
plt.title(f'Boxplot de JamsDelay (2024)\nPromedio: {mean_value:.2f}')
plt.ylabel('Minutos de congestión (JamsDelay)')
plt.show()
```

```python
# Crear histograma para ver la distribución de la economía (city_gdp_capita)
sns.histplot(merged['city_gdp_capita'], kde=True, bins=10)

plt.title('Distribución del PIB per cápita (2024)')
plt.xlabel('PIB per cápita (City GDP/capita)')
plt.ylabel('Frecuencia / Número de ciudades')
```

```python
# Gráfico de barras para comparar jams_delay y city_gdp_capita por ciudad
merged.plot(x='city', y=['JamsDelay', 'city_gdp_capita'],
    kind='bar',
    figsize=(10, 5),
)
plt.title('Comparación de Congestión Vial y PIB per cápita por Ciudad (2024)')
plt.xlabel('Ciudad')
plt.ylabel('Valor de Indicadores')
plt.xticks(rotation=90)
plt.show()
```

### 🧠 **Hallazgo**
No existe una relación directa ni lineal clara entre el PIB per cápita y la congestión vial.

Aunque existe una ligera tendencia positiva general (un PIB más alto suele ir acompañado de más vehículos en circulación), los datos demuestran que un mayor PIB per cápita no garantiza mayor congestión, ni viceversa.


```python
# Exporta el dataset final como CSV
merged.to_csv("ladb_mobility_economy_2024_clean.csv", index=False)
```

# 🧾 Resumen ejecutivo


**Contexto & objetivo:**  
El objetivo central de este análisis es evaluar cómo los niveles de movilidad urbana (congestión vehicular, tiempos de viaje y demoras por embotellamiento) impactan la productividad económica de las ciudades latinoamericanas. Para ello se examinaron variables clave como el retraso total por embotellamiento (JamsDelay), el tiempo de viaje promedio por 10 km (TravelTimeLivePer10KmsMins), el PIB per cápita metropolitano (city_gdp_capita) y la tasa de desempleo (unemployment_pct). La evaluación de este cruce de variables es fundamental para orientar decisiones estratégicas de inversión para aumentar la productividad y bienestar de la población.

**Cobertura de datos:**  
El dataset consolidado comprende datos del año 2024 para 15 ciudades latinoamericanas pertenecientes a 7 países (Argentina, Brasil, Chile, Colombia, México, Perú y Uruguay). A nivel metodológico, se estandarizaron los nombres de las columnas a formato snake_case, se limpiaron caracteres especiales ($, %, ,, .) en los campos económicos y se realizó una conversión de tipos de datos a flotantes y fechas. Los datos de tráfico de TomTom fueron agregados mediante sus promedios anuales por ciudad y posteriormente combinados con los datos macroeconómicos de la OECD mediante una unión de tipo INNER JOIN usando las llaves de ciudad y año (city y year).

Los resultados revelan que la congestión vehicular está directamente ligada al tamaño poblacional, observándose valores atípicos (outliers) severos en grandes ciudades con gran población como en Ciudad de México ($2,833.06$ min de demoras promedio). Si bien a nivel global las ciudades con mayor PIB per cápita tienden a concentrar mayor actividad vehicular, se identificó un importante desajuste o ineficiencia estructural en ciertas capitales, donde altos índices de embotellamiento conviven con un PIB per cápita moderado o bajo.


**Recomendaciones**  
Bogotá (Colombia) se posiciona como la ciudad prioritaria para inversión inmediata en infraestructura de transporte. Al comparar el grupo de megalópolis de la región, Bogotá presenta un tiempo de viaje crítico ($24.99$ min para recorrer 10 km, el segundo más alto del estudio solo detrás de Lima con $25.67$ min) y una demora promedio severa ($1,141.55$ min de JamsDelay), mientras mantiene un PIB per cápita relativamente bajo ($\$11,442$ USD) y la mayor tasa de desempleo del grupo primario ($10.0\%$). A diferencia de Buenos Aires ($\$18,117$ USD PIB/cápita) o Ciudad de México ($\$21,111$ USD PIB/cápita), en Bogotá las fricciones de movilidad actúan como un cuello de botella directo que frena la productividad económica, haciendo urgente la expansión de sistemas masivos de transporte
